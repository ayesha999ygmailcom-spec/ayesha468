# Data Extraction and Inspection
Let's begin by extracting the contents of the uploaded ZIP file to see what datasets are available for analysis.

```python
import zipfile
import os

zip_path = '/content/archive (3).zip'
extract_path = '/content/extracted_data'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Extracted files:")
for root, dirs, files in os.walk(extract_path):
    for file in files:
        print(os.path.join(root, file))
```

# Data Loading and Basic Inspection
Let's load the `starbucks.csv` file into a Pandas DataFrame and check its dimensions, columns, data types, and any missing values.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/extracted_data/starbucks.csv')

# Display basic details
print("Dataset Shape:", df.shape)
print("\n--- Dataset Columns & Types ---")
df.info()

print("\n--- First 5 Rows ---")
display(df.head())

print("\n--- Missing Values Summary ---")
print(df.isnull().sum())
```

# Exploratory Data Analysis & Visualizations
Let's generate statistical summaries and plots to better understand the distributions of numerical columns and any relationships within the data.

```python
# Clean column names by stripping leading/trailing whitespaces
df.columns = df.columns.str.strip()

# Statistical summary of numeric columns
print("\n--- Descriptive Statistics ---")
display(df.describe(include='all'))

# Visualizing distributions of key numerical attributes if present
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if len(numeric_cols) > 0:
    plt.figure(figsize=(12, 8))
    for i, col in enumerate(numeric_cols[:6], 1):
        plt.subplot(2, 3, i)
        sns.histplot(df[col].dropna(), kde=True, bins=20)
        plt.title(f'Distribution of {col}')
    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns found for distribution plots.")
```

# Model Preparation: Train-Test Split & Decision Tree
Now we will prepare the dataset for predictive modeling. We will split the dataset into 80% training and 20% testing data, fit a Decision Tree, and evaluate its performance.

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.metrics import classification_report, accuracy_score, mean_squared_error, r2_score
from sklearn.preprocessing import LabelEncoder

# Prepare copy of dataframe
model_df = df.copy()

# Automatically identify categorical columns and encode them
label_encoders = {}
for col in model_df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    model_df[col] = le.fit_transform(model_df[col].astype(str))
    label_encoders[col] = le

# Let's dynamically select the target column. We'll check if a common column like 'Calories' exists to predict, or use the last numeric column.
if 'Calories' in df.columns:
    target_col = 'Calories'
elif len(numeric_cols) > 0:
    target_col = numeric_cols[-1]
else:
    target_col = model_df.columns[-1]

print(f"Selected target column for modeling: {target_col}")

X = model_df.drop(columns=[target_col])
y = model_df[target_col]

# Train-test split (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Initialize and fit the Decision Tree depending on whether the target is continuous or discrete
if y.nunique() > 10 and np.issubdtype(y.dtype, np.number):
    print("Running Decision Tree Regressor...")
    model = DecisionTreeRegressor(random_state=42, max_depth=5)
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    print("Mean Squared Error:", mean_squared_error(y_test, predictions))
    print("R2 Score:", r2_score(y_test, predictions))
else:
    print("Running Decision Tree Classifier...")
    model = DecisionTreeClassifier(random_state=42, max_depth=5)
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    print("Accuracy:", accuracy_score(y_test, predictions))
    print("\nClassification Report:\n", classification_report(y_test, predictions))
```

# Model Evaluation & Post-Testing Analysis
Let's visualize the results of our Decision Tree model by plotting actual vs. predicted values and examining feature importances.

```python
# 1. Plot Actual vs. Predicted Calories
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.scatterplot(x=y_test, y=predictions, alpha=0.7, color='teal')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.title('Actual vs. Predicted Calories')
plt.xlabel('Actual Calories')
plt.ylabel('Predicted Calories')
plt.grid(True, linestyle='--', alpha=0.6)

# 2. Plot Feature Importances
importances = model.feature_importances_
indices = np.argsort(importances)[::-1]
feature_names = X.columns

plt.subplot(1, 2, 2)
sns.barplot(x=importances[indices], y=feature_names[indices], palette='viridis')
plt.title('Feature Importances in Predicting Calories')
plt.xlabel('Relative Importance')
plt.ylabel('Features')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# Regression Performance Metrics (Accuracy & Precision equivalents)
Let's compute the Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE) to quantify the prediction errors in terms of calorie units.

```python
from sklearn.metrics import mean_absolute_error

# Calculate regression equivalents of accuracy and precision
mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print("--- Model Prediction Quality Metrics ---")
print(f"Mean Absolute Error (MAE): {mae:.2f} calories")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f} calories")
print(f"R2 Score (Variance Explained): {r2 * 100:.2f}%")
```

# Visualizing Prediction Accuracy with a Pie Chart
Let's calculate the absolute errors for the test set predictions, group them into descriptive accuracy ranges, and visualize them using a pie chart.

```python
# Calculate absolute errors
absolute_errors = np.abs(y_test - predictions)

# Categorize errors
bins = [0, 10, 30, np.inf]
labels = ['Highly Accurate (Error ≤ 10 kcal)', 'Moderately Accurate (Error 10-30 kcal)', 'Lower Accuracy (Error > 30 kcal)']
error_categories = pd.cut(absolute_errors, bins=bins, labels=labels, include_lowest=True)
category_counts = error_categories.value_counts()

# Plot pie chart
plt.figure(figsize=(8, 8))
plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=['#2ca02c', '#ff7f0e', '#d62728'],
    explode=(0.05, 0.05, 0.05)
)
plt.title('Distribution of Model Prediction Accuracy (Absolute Errors)')
plt.show()
```

### Sample Predictions Table
Let's extract a sample of 10 drinks from the test set, map their encoded variables back to their readable names, and display their actual vs. predicted calorie values.

```python
# Select 10 samples from the test set
sample_indices = X_test.head(10).index

# Create a comparison dataframe
sample_comparison = pd.DataFrame(index=sample_indices)

# Decode Beverage and Beverage_prep names using the fitted LabelEncoders
sample_comparison['Beverage'] = label_encoders['Beverage'].inverse_transform(X_test.loc[sample_indices, 'Beverage'])
sample_comparison['Beverage_prep'] = label_encoders['Beverage_prep'].inverse_transform(X_test.loc[sample_indices, 'Beverage_prep'])

# Extract actual and predicted values
sample_comparison['Actual Calories'] = y_test.loc[sample_indices]
sample_comparison['Predicted Calories'] = np.round(predictions[:10], 1) # matches head(10) mapping
sample_comparison['Absolute Error'] = np.abs(sample_comparison['Actual Calories'] - sample_comparison['Predicted Calories'])

display(sample_comparison)
```

### Saving the Model
We will serialize and save our trained model and its associated label encoders into a pickle file (`starbucks_calorie_model.pkl`) for future deployment and inference.

```python
import pickle

# Bundle the model and encoders together
model_artifacts = {
    'model': model,
    'label_encoders': label_encoders
}

# Save to a pickle file
model_filename = '/content/starbucks_calorie_model.pkl'
with open(model_filename, 'wb') as file:
    pickle.dump(model_artifacts, file)

print(f"Successfully saved model and encoders to: {model_filename}")
```
